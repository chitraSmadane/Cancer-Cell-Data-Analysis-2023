```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('All libraries loaded successfully!')
```

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df['diagnosis'] = df['target'].map({1: 'Benign', 0: 'Malignant'})

print("Dataset loaded!")
print("Shape:", df.shape)
df.head()
```

```python
# How many Benign and Malignant cases?
print("Diagnosis Count:")
print(df['diagnosis'].value_counts())

sns.countplot(x='diagnosis', data=df, palette='Set2')
plt.title('Benign vs Malignant Cases')
plt.xlabel('Diagnosis')
plt.ylabel('Count')
plt.show()
```

```python
# Average values for Benign vs Malignant
df.groupby('diagnosis')[['mean radius', 'mean texture', 
'mean perimeter', 'mean area']].mean().plot(kind='bar', 
figsize=(10,5), colormap='Set2')

plt.title('Average Feature Values: Benign vs Malignant')
plt.xlabel('Diagnosis')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(12,8))
sns.heatmap(df[['mean radius', 'mean texture', 'mean perimeter', 
'mean area', 'mean smoothness']].corr(), 
annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Between Features')
plt.tight_layout()
plt.show()
```

```python
# Boxplot - tumor size distribution
plt.figure(figsize=(8,5))
sns.boxplot(x='diagnosis', y='mean radius', 
data=df, palette='Set2')
plt.title('Tumor Size Distribution: Benign vs Malignant')
plt.xlabel('Diagnosis')
plt.ylabel('Mean Radius')
plt.show()
```
